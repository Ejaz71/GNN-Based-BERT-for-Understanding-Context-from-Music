# Task 1 EDA: MusicCaps captions & aspect tags

Explores the `google/MusicCaps` dataset before training: caption length distribution and the frequency of the top-K aspect tags used as multi-label targets.

```python
import ast
import json
import sys
from collections import Counter
from pathlib import Path

sys.path.append(str(Path.cwd().parent))
import matplotlib.pyplot as plt
from datasets import load_dataset

from src.utils import ROOT, load_config

cfg = load_config()
dataset = load_dataset(cfg["data"]["dataset_name"], split="train", cache_dir=str(ROOT / cfg["data"]["raw_dir"]))
print(f"Rows: {len(dataset)}")
```

```python
caption_lens = [len(c.split()) for c in dataset["caption"]]
plt.figure()
plt.hist(caption_lens, bins=30)
plt.xlabel("Caption length (words)")
plt.ylabel("Count")
plt.title("MusicCaps caption length distribution")
plt.show()
print("mean:", sum(caption_lens) / len(caption_lens), "max:", max(caption_lens))
```

```python
counter = Counter()
for row in dataset["aspect_list"]:
    counter.update(ast.literal_eval(row))
print("unique aspects:", len(counter))

top50 = counter.most_common(cfg["data"]["top_k_tags"])
tags, counts = zip(*top50)
plt.figure(figsize=(10, 12))
plt.barh(range(len(tags)), counts)
plt.yticks(range(len(tags)), tags, fontsize=8)
plt.gca().invert_yaxis()
plt.xlabel("Frequency")
plt.title(f"Top {cfg['data']['top_k_tags']} aspect tags used as Task 1 labels")
plt.tight_layout()
plt.show()
```
